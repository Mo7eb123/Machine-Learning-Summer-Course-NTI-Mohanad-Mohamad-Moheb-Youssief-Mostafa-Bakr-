```python

# 1. Calculate percentage cross-tabulation
ct = (
    pd.crosstab(
        df['Major_Category'], df['Burnout_Risk_Level'], normalize='index'
    )[['Low', 'Medium', 'High']]
    * 100
)

# 2. Plot stacked bar chart directly
ax = ct.plot(kind='bar', stacked=True, figsize=(9, 5), rot=0)

# 3. Add percentage labels using fmt string
for container in ax.containers:
  ax.bar_label(
      container,
      fmt='%.2f%%',
      label_type='center',
      fontsize=9,
      fontweight='bold',
      color='white',
  )

plt.title('Proportion of Burnout Risk Levels Across Major Categories')
plt.ylabel('Percentage (%)')
plt.legend(title='Burnout Risk', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.show()
```

* STM  major has the highest high burnout risks

```python
cats = ['Institutional_Policy', 'Prompt_Engineering_Skill', 'Primary_Use_Case', 'Major_Category']
fig, axes = plt.subplots(2, 2, figsize=(18, 14))

for col, ax in zip(cats, axes.flatten()):
    # Calculate row-wise proportions (%)
    ct = pd.crosstab(df[col], df['Burnout_Risk_Level'], normalize='index')[['Low', 'Medium', 'High']] * 100

    # Plot stacked bar chart
    ct.plot(kind='bar', stacked=True, ax=ax, colormap='coolwarm', legend=(ax == axes[0, 0]))

    # Write percentage values inside each box
    for c in ax.containers:
        # Option A: Standard formatting for all boxes
        # ax.bar_label(c, fmt='%.1f%%', label_type='center', fontsize=9, fontweight='bold')

        # Option B: Clean formatting (hides percentage if segment is smaller than 2% to prevent overlap)
        labels = [f'{height:.1f}%' if (height := rect.get_height()) >= 2 else '' for rect in c]
        ax.bar_label(c, labels=labels, label_type='center', fontsize=9, fontweight='bold', color='black')

    ax.set_title(f'Burnout Risk Density by {col}', fontsize=12, fontweight='bold')
    ax.set_ylabel('Percentage (%)')
    ax.set_xlabel('')
    ax.set_ylim(0, 105)  # Keeps bars nicely within bounds
    ax.tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

# splitting and scaling the model

```python
from sklearn.model_selection import train_test_split
class_x = df.drop(columns="Burnout_Risk_Level")
class_y = df["Burnout_Risk_Level"]

class_bool_cols = class_x.select_dtypes(include="bool").columns

class_cols_order = class_x.columns

x_train, x_test, y_train, y_test = train_test_split(class_x, class_y, test_size=0.3, random_state = 58)
```

```python
print(class_nums)
print(class_cats)
print(class_cols_order)
```

```python
from sklearn.preprocessing import StandardScaler


# 1. Instantiate once
class_scaler = StandardScaler()

# 2. Fit and transform training data on 'scaler'
x_train[numerical_data] = class_scaler.fit_transform(x_train[numerical_data])

# 3. Transform test data using the SAME 'scaler' instance
x_test[numerical_data] = class_scaler.transform(x_test[numerical_data])

x_train.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
from sklearn.naive_bayes import BernoulliNB
from sklearn.ensemble import AdaBoostClassifier , RandomForestClassifier
```

```python
class_models={"LR": LogisticRegression(max_iter=1000),"Tree":DecisionTreeClassifier(),"RF":RandomForestClassifier(),"KNN":KNeighborsClassifier(),'ADABOOST':AdaBoostClassifier()}
```

```python
for name ,model in class_models.items():
    model.fit(x_train,y_train)
    y_pre_train = model.predict(x_train)
    y_pre_test = model.predict(x_test)
    print(name)
    print(classification_report(y_train,y_pre_train))
    print(classification_report(y_test,y_pre_test))
    print("------------------------------------------------")
```

```python
x_train, x_test, y_train, y_test = train_test_split(reg_x, reg_y, test_size=0.3, random_state = 58)
```

```python
reg_numerical_data = numerical_data.drop(["Post_Semester_GPA","GPA_Delta"])
# 1. Instantiate once
reg_scaler =StandardScaler()

# 2. Fit and transform training data on 'scaler'
x_train[reg_numerical_data] = reg_scaler.fit_transform(x_train[reg_numerical_data])

# 3. Transform test data using the SAME 'scaler' instance
x_test[reg_numerical_data] = reg_scaler.transform(x_test[reg_numerical_data])

x_train.head()
```

```python
from sklearn.ensemble import AdaBoostRegressor, RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
```

```python
Regression_models = {
    'LR': LinearRegression(),
    'Tree': DecisionTreeRegressor(random_state=42),
    'RF': RandomForestRegressor(random_state=42),
    'KNN': KNeighborsRegressor(),
    'ADABOOST': AdaBoostRegressor(random_state=42),
}
```

```python
for name, model in Regression_models.items():
  model.fit(x_train, y_train)

  y_pred_train = model.predict(x_train)
  y_pred_test = model.predict(x_test)

  print(f'================ {name} ================')

  # Train metrics
  train_r2 = r2_score(y_train, y_pred_train)
  train_rmse = np.sqrt(mean_squared_error(y_train, y_pred_train))
  train_mae = mean_absolute_error(y_train, y_pred_train)

  # Test metrics
  test_r2 = r2_score(y_test, y_pred_test)
  test_rmse = np.sqrt(mean_squared_error(y_test, y_pred_test))
  test_mae = mean_absolute_error(y_test, y_pred_test)

  print(
      f'TRAIN -> R2: {train_r2:.4f} | RMSE: {train_rmse:.4f} | MAE:'
      f' {train_mae:.4f}'
  )
  print(
      f'TEST  -> R2: {test_r2:.4f} | RMSE: {test_rmse:.4f} | MAE:'
      f' {test_mae:.4f}'
  )
  print('------------------------------------------------\n')
```
